## Matplotlib example workflow

```python
# A matplotlib workflow

# 0. Import and get matplotlib ready
import matplotlib.pyplot as plt

# 1. Prepare / Get your data
x= [1,2,3,4]
y=[11,22,33,44]

# 2. Setup plot ( instances of the objects)
fig,ax = plt.subplots(figsize=(10,10)) # 

# 3. Plot data
ax.plot(x,y)

# 4. Customize plot
ax.set(title="Simple Sample plot" , 
       xlabel="X-Axis",
       ylabel="Y-Axis")

# 5. Save & Show
fig.savefig("numpy_images/simple-sample-plot.png")

       
```

## Making figures with Numpy arrays

Most of figuring out what kind of plot to use is getting a feel for the data, then see what suits it best.

Matplotlib visualizations are built off NumPy arrays. So in this section we'll build some of the most common types of plots using NumPy arrays.

* line plot (default)
* scatter plot
* bar plot
* histogram
* subplots()

# Subplots
 * We will explore two different ways to make subplots

```python
# Option 1 : Create multiple subplots
fig, ((ax1,ax2), (ax3,ax4))  = plt.subplots(nrows=2,
                                            ncols=2,
                                            figsize=(10,5))

# Now create values to view inside each sub plot
ax1.plot(x,x/2)
ax2.scatter(np.random.rand(10), np.random.rand(10))
ax3.bar(sanjiv_fav_food_prices.keys(), sanjiv_fav_food_prices.values())
ax4.hist(np.random.randn(3000)); ## the semicolon will remove the random gibirish that comes wiht output (remove it and run to see what happens)
```

```python
# Option 2 : Different way to plot using indexing
fig, ax = plt.subplots(nrows=2,
                        ncols=2,
                      figsize=(10,5))
## Now use indexing to access each subplot
ax[0,0].plot(x,x/2)
ax[0,1].scatter(np.random.random(10),np.random.rand(10))
ax[1,0].barh(sanjiv_fav_food_prices.keys(), sanjiv_fav_food_prices.values())
ax[1,1].hist(np.random.randn(2500));
```

### Which one should i use ? (pyplot vs matplotlib object oriented method)
* When plotting something quickly and simple plots, ok to use pyplot
* In all the other cases, use matplotlib for visibility and clearness

```python
# Perform Data Analysis on patients over 50
over_50 = heart_disease[heart_disease["age"] > 50]
over_50
```

```python
len(over_50)
```

```python
# Pyplot method aka plotting directly using pandas dataframe method
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target", # this is basically for colour coding the output
             figsize=(10,6))
```

```python
# Using the matplotlib OO method, kinda (OO + pyplot)
fig,ax = plt.subplots(figsize=(10,6))
over_50.plot(kind="scatter",
             x='age',
             y='chol',
             c='target',
             ax=ax)

# ax.set_xlim([45,100])
# the above line is to basically alter the x axis length so as to view our plots more accurately and clearly as per our will
```

```python
over_50
```

```python
## OO method from scratch

# Create the object
fig,ax = plt.subplots(figsize=(10,6))

# Plot the data 
scatter = ax.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"])

# Customize the plot (add labels basically)
ax.set(title="Heart Disease and Cholestrol Level",
       xlabel="Age",
       ylabel="Cholestrol")

# Legend addition (VIMP)
ax.legend(*scatter.legend_elements(), title="Legend")


# Add a mean line
ax.axhline(over_50["chol"].mean(),
          linestyle="--");
```

```python
over_50.head()
```

```python
## Add a subplot, this shows the full power of OO method, the ability to customize our own plots
```

```python
# Create a OO object instance
fig, (ax0,ax1)  = plt.subplots(nrows=2,
                               ncols=1,
                               sharex=True, # -> this is if you want both the subplots to share the x-axis
                              figsize=(10,10))

# Add data to the ax0 subplot
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"])

# Add labels to ax0
ax0.set(title="Heart Disease and Cholestrol Levels Comparision",
        # xlabel="Age",  #-> this is not required when we add 'sharex' parameter on the instance declaration
        ylabel="Cholesterol Level")

# Add a legend to ax0
ax0.legend(*scatter.legend_elements(),title="Legend")

# Add a mean line to ax0
ax0.axhline(over_50["chol"].mean(),
            linestyle="--");

#---------------------------------------------------------------------------------------------------------

# Add data to the ax1 subplot

scatter = ax1.scatter(x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"])

# Add labels to the ax1 subplot

ax1.set(title = "Heart Disease and Max Heart rate comparision",
       xlabel="Age",
        ylabel="Max Heart Rate")

# Add legend to ax1 subplot

ax1.legend(*scatter.legend_elements(), title="Legend")


# Add a mean line to ax1

ax1.axhline(over_50["thalach"].mean(),
           linestyle="dotted")

#-----------------------------------------------------------------------------------------------------------

# Add a Tittle to the whole figure

fig.suptitle("Heart Disease Analysis against Cholestrol and Max heart rate", fontsize=20, fontweight='extra bold', fontfamily='serif',color="Red");
```

## Some other changes to subplots

```python
# Change cmap and horizontal line to be a different colour
fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(over_50["age"], 
                     over_50["chol"], 
                     c=over_50["target"], 
                     cmap="winter") ## -> this is what we change, just changes the colour scheme, research official doc for more schemes
ax.set(title="Heart Disease and Cholesterol Levels",
       xlabel="Age",
       ylabel="Cholesterol")
ax.axhline(y=over_50["chol"].mean(), 
           color='r', ## -> this changes colour of our horizontal line
           linestyle='--', 
           label="Average");
ax.legend(*scatter.legend_elements(), title="Target");
```

# Changing the xlim and ylim to ensure that the plots dont arbitarily start or cut of at edges and its clean

```python
## Before the change , we can see lot of cutt-offs , now we will change the xlim and ylim below the set() methods in the next cell

fig, (ax0, ax1) = plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(10, 10))
scatter = ax0.scatter(over_50["age"], 
                      over_50["chol"], 
                      c=over_50["target"],
                      cmap='winter')
ax0.set(title="Heart Disease and Cholesterol Levels",
        ylabel="Cholesterol")

# Setup a mean line
ax0.axhline(y=over_50["chol"].mean(), 
            color='r', 
            linestyle='--', 
            label="Average");
ax0.legend(*scatter.legend_elements(), title="Target")

# Axis 1, 1 (row 1, column 1)
scatter = ax1.scatter(over_50["age"], 
                      over_50["thalach"], 
                      c=over_50["target"],
                      cmap='winter')
ax1.set(title="Heart Disease and Max Heart Rate Levels",
        xlabel="Age",
        ylabel="Max Heart Rate")

# Setup a mean line
ax1.axhline(y=over_50["thalach"].mean(), 
            color='r', 
            linestyle='--', 
            label="Average");
ax1.legend(*scatter.legend_elements(), title="Target")

# Title the figure
fig.suptitle('Heart Disease Analysis', fontsize=16, fontweight='bold');
```

```python
## Before the change (we've had color updates)

fig, (ax0, ax1) = plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(10, 20))
scatter = ax0.scatter(over_50["age"], 
                      over_50["chol"], 
                      c=over_50["target"],
                      cmap='winter')
ax0.set(title="Heart Disease and Cholesterol Levels",
        ylabel="Cholesterol")

#------- Adding this extra-------------------------------------
ax0.set_xlim([50,80])

# Setup a mean line
ax0.axhline(y=over_50["chol"].mean(), 
            color='r', 
            linestyle='--', 
            label="Average");
ax0.legend(*scatter.legend_elements(), title="Target")

# Axis 1, 1 (row 1, column 1)
scatter = ax1.scatter(over_50["age"], 
                      over_50["thalach"], 
                      c=over_50["target"],
                      cmap='winter')
ax1.set(title="Heart Disease and Max Heart Rate Levels",
        xlabel="Age",
        ylabel="Max Heart Rate")

# --------------Adding this extra ----------------------------
ax1.set_ylim([60,200])

# Setup a mean line
ax1.axhline(y=over_50["thalach"].mean(), 
            color='r', 
            linestyle='--', 
            label="Average");
ax1.legend(*scatter.legend_elements(), title="Target")

# Title the figure
fig.suptitle('Heart Disease Analysis', fontsize=16, fontweight='bold')

#-------------------------------------------------
## SOMETIMES, THE FIGURE TITLE OVERLAPS WITH THE SUBPLOTS TITLE , TO AVOID THAT, WE ADD THE BELOW LINE
plt.tight_layout(rect=[0, 0, 1, 0.95])  # <- Add this line
```
